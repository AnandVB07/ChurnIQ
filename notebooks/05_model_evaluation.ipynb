# 05. Model Evaluation & Error Diagnostics

## Key Diagnostics
- Precision, Recall, F1, ROC-AUC, and Brier Score
- Confusion Matrix breakdown (False Positives vs False Negatives)
- Calibration and Business Tradeoffs

```python
import json
import seaborn as sns
import matplotlib.pyplot as plt
from src.config import MODEL_METADATA_FILE

with open(MODEL_METADATA_FILE, 'r') as f:
    metadata = json.load(f)

print(f"Model: {metadata['model_name']} (Version {metadata['version']})")
print("Metrics:", json.dumps(metadata['metrics'], indent=2))

cm = metadata['confusion_matrix']['matrix']
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Purples', xticklabels=['Stay', 'Churn'], yticklabels=['Stay', 'Churn'])
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.title('Test Set Confusion Matrix')
plt.show()
```
