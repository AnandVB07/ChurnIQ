# 02. Exploratory Data Analysis (EDA)

## Key Investigations
1. Overall Churn Rate & Class Balance
2. Churn rate stratified across Contract Types (Month-to-month vs Annual)
3. Churn behavior by Tenure and Monthly Charges
4. Impact of Tech Support, Online Security, and Internet Provider
5. Payment Method & Billing Preferences

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from src.data_loader import load_or_download_data
from src.preprocessing import clean_raw_data

df = clean_raw_data(load_or_download_data())
sns.set_theme(style="darkgrid")
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(x='Contract', hue='Churn', data=df, palette='Purples', ax=ax[0])
ax[0].set_title('Churn Count by Contract Type')

sns.kdeplot(data=df, x='MonthlyCharges', hue='Churn', fill=True, common_norm=False, palette='coolwarm', ax=ax[1])
ax[1].set_title('Monthly Charges Distribution by Churn')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x='PaymentMethod', y='tenure', hue='Churn', data=df, palette='Purples')
plt.xticks(rotation=20)
plt.title('Tenure Distribution by Payment Method and Churn Status')
plt.show()
```
