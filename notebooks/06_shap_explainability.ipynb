```python
from src.predict import ChurnPredictor
import pandas as pd

predictor = ChurnPredictor()
sample_customer = {
    "gender": "Female",
    "SeniorCitizen": 0,
    "Partner": "No",
    "Dependents": "No",
    "tenure": 2,
    "PhoneService": "Yes",
    "MultipleLines": "No",
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "No",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "Yes",
    "StreamingMovies": "Yes",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 92.50,
    "TotalCharges": 185.00
}

result = predictor.predict(sample_customer)
print(f"Churn Probability: {result['churn_probability'] * 100:.1f}%")
print(f"Risk Level: {result['risk_level']}")
print("\nTop SHAP Risk Factors:")
for f in result['top_factors']:
    print(f"  • {f['feature_label']}: {f['shap_value']:+.3f} ({f['impact']})")

print("\nRetention Recommendation:")
print("Summary:", result['recommendation']['summary'])
print("Actions:", result['recommendation']['action_items'])
```
